# data reading and writing and creatling delta tables


```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import *
from pyspark import *
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pyspark.sql.functions import split, col
```

```python
%sql
create  database gold
```

# Data Zone



```python
df_zone = (
    spark.read
    .format("parquet")
    .option("header", True)
    .option("inferSchema", True)
    .load("abfss://silver@nyctaxideepa.dfs.core.windows.net/trip_zone")
)
```

```python
df_zone.display()
```

```python
df_zone.write.format("delta") \
    .mode("append") \
    .option("path", "abfss://gold@nyctaxideepa.dfs.core.windows.net/trip_zone") \
    .saveAsTable("gold.trip_zone")
```

```python
%sql
select * from gold.trip_zone
```
