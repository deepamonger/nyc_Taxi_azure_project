Trip zone


```python
df_trip_type = (
    spark.read
    .format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load("abfss://bronze@nyctaxideepa.dfs.core.windows.net/taxi_zone_lookup.csv")
)

display(df_trip_type)
```

```python
df_trip = spark.read.format("parquet") \
    .schema(myschema) \
    .option("header", True) \
    .option("recursiveFileLookup", True) \
    .load("abfss://bronze@nyctaxideepa.dfs.core.windows.net/trip-data/")
```

```python
df_trip.display()
```

```python
myschema = '''
    VendorID BIGINT,
    lpep_pickup_datetime TIMESTAMP,
    lpep_dropoff_datetime TIMESTAMP,
    store_and_fwd_flag STRING,
    RatecodeID BIGINT,
    PULocationID BIGINT,
    DOLocationID BIGINT,
    passenger_count BIGINT,
    trip_distance DOUBLE,
    fare_amount DOUBLE,
    extra DOUBLE,
    mta_tax DOUBLE,
    tip_amount DOUBLE,
    tolls_amount DOUBLE,
    ehail_fee DOUBLE,
    improvement_surcharge DOUBLE,
    total_amount DOUBLE,
    payment_type BIGINT,
    trip_type BIGINT,
    congestion_surcharge DOUBLE
'''
```

```python
df_trip.display()
```

**Data Tranformation**

```python
display(df_trip_type)
```

```python
f_trip = df_trip_type.withColumnRenamed("Description", "New_Name")
```

```python
display(f_trip)
```

```python
df_trip_type.write.format("parquet") \
    .mode("append") \
    .option(
        "path",
        "abfss://silver@nyctaxideepa.dfs.core.windows.net/trip_type"
    ) \
    .save()
```

```python
df_trip_type.columns
```

```python
from pyspark.sql.functions import split, col, get, lit

df_trip_type = (
    df_trip_type
    .withColumn("zone1", get(split(col("Zone"), "/"), lit(0)))
    .withColumn("zone2", get(split(col("Zone"), "/"), lit(1)))
)

display(df_trip_type)
```

```python
df_trip_type.write \
    .format("parquet") \
    .mode("append") \
    .option(
        "path",
        "abfss://silver@nyctaxideepa.dfs.core.windows.net/df_trip_type"
    ) \
    .save()
```

```python
from pyspark.sql.functions import to_date, year, month

df_trip = (
    df_trip
    .withColumn("trip_date", to_date("lpep_pickup_datetime"))
    .withColumn("trip_year", year("lpep_pickup_datetime"))
    .withColumn("trip_month", month("lpep_pickup_datetime"))
)

display(df_trip)
```
